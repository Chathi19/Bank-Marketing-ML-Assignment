```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Performance results from both models
results = {
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-score'],
    'Logistic Regression': [0.8917, 0.5536, 0.2981, 0.3875],
    'Decision Tree': [0.8895, 0.5312, 0.3269, 0.4048]
}

# Convert results into a DataFrame
comparison_df = pd.DataFrame(results)

# Display the comparison table
comparison_df
```

```python
# Create a bar chart comparing the two models
comparison_df.set_index('Metric').plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Logistic Regression vs Decision Tree')
plt.xlabel('Evaluation Metrics')
plt.ylabel('Score')
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title='Algorithms')
plt.tight_layout()
plt.show()
```

## Model Comparison and Conclusion

Two supervised machine learning algorithms, Logistic Regression and Decision Tree, were implemented using the Bank Marketing dataset.

Logistic Regression achieved 89.17% accuracy, while Decision Tree achieved 88.95% accuracy. Logistic Regression performed slightly better in terms of accuracy and precision.

However, Decision Tree achieved higher recall (32.69%) and F1-score (40.48%) compared with Logistic Regression, which achieved 29.81% recall and 38.75% F1-score.

The dataset is imbalanced, with approximately 88.5% of customers not subscribing to a term deposit. Therefore, accuracy alone is not sufficient to evaluate model performance.

In conclusion, Decision Tree performed slightly better in identifying customers who subscribed, while Logistic Regression achieved slightly higher overall accuracy. Both models could be improved, particularly in detecting subscribing customers.

One limitation is that the duration feature is only known after a marketing call. Therefore, these results should not be interpreted as predictions that can be made before contacting a customer.
